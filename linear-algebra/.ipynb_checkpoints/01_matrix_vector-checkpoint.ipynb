```python
def matvec(A,x):
    result = [0] * len(A)
    for i in range(len(A[0])):
        for j in range (len(A)):
            result [j] += A[j][i] * x[i]
    return result
            
matvec([[0, -1], [1, 0]],[2, 1])
    
```

```python
import numpy as np
def numpyvec(A,x):
    return np.array(A) @ np.array(x)
numpyvec([[0, -1], [1, 0]],[2, 1])
    
```

```python
def check(A,x):
    matvec_result = matvec(A,x)
    numpyvec_result = numpyvec(A,x)
    assert np.array_equal(matvec_result, numpyvec_result), "Mat vec is not correct"
    print(matvec_result, numpyvec_result)

check([[0, -1], [1, 0]],[2, 1])
check([[1, 1], [0, 1]], [1, 2])
check([[2, -1], [-1, 2]], [1, 2])

    
```

```python
import matplotlib.pyplot as plt

def vec_plot (points):
    shear = [[0, -1], [1, 0]]
    old_xs= [p[0] for p in points]
    old_ys= [p[1] for p in points]
    new_points = [matvec(shear,p) for p in points]
    new_xs = [p[0] for p in new_points]
    new_ys = [p[1] for p in new_points]

    plt.plot(old_xs, old_ys)
    plt.plot(new_xs, new_ys, "r")
    plt.axis("equal")
    plt.grid(True)
    plt.show()
    
    
points = [[0, 0], [1, 0], [1, 1], [0, 1],[0,0]]
vec_plot(points)
 
```
